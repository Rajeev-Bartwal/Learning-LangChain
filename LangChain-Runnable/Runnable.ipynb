# Runnables in LangChain

## 1. What is a Runnable?

A **Runnable** is a common abstraction/interface in LangChain that defines how an executable component should behave.

Different LangChain components can implement the `Runnable` interface and provide their own internal logic.

Examples:

```text
PromptTemplate
ChatModel
OutputParser
Retriever
RunnableLambda
```

These components may perform completely different tasks, but they follow a **common Runnable interface**.

### Simple idea

```text
             Runnable
          Common Interface
                │
      ┌─────────┼─────────┐
      ↓         ↓         ↓
   Prompt     Model     Parser
   Runnable   Runnable  Runnable
```

Each component does its own work internally, but exposes a standardized way of being executed.

---

# 2. Why Runnable?

Before understanding Runnable, think about the problem.

Suppose we have:

```text
Prompt → Model → Parser
```

These are different types of components.

The Prompt creates a prompt.

The Model generates a response.

The Parser converts the response into structured data.

Internally, all three work differently.

The problem is:

> **How can we make these different components work together in a standardized way?**

LangChain solves this using the **Runnable abstraction**.

---

# 3. Common Interface

Conceptually, imagine LangChain defines:

```python
class Runnable:

    def invoke(self, input):
        pass
```

Now different components can follow this common contract.

For example:

```python
class Prompt(Runnable):

    def invoke(self, input):
        # prompt-specific logic
        ...


class Model(Runnable):

    def invoke(self, input):
        # model-specific logic
        ...


class Parser(Runnable):

    def invoke(self, input):
        # parser-specific logic
        ...
```

Notice something important:

The **internal implementation is different**.

But externally, they provide a common execution interface.

```text
Prompt  ──┐
Model    ──┼──> common Runnable interface
Parser   ──┘
```

---

# 4. Standardization

Because components follow the Runnable abstraction, we can interact with them in a standardized way.

For example:

```python
prompt.invoke(input)
```

```python
model.invoke(input)
```

```python
parser.invoke(input)
```

The internal implementation is different, but the way we execute them is standardized.

---

# 5. Composition

This is where Runnable becomes really powerful.

Because these components follow a common interface, they can be **composed together**.

For example:

```python
chain = prompt | model | parser
```

This creates a sequence:

```text
User Input
    ↓
 Prompt
    ↓
  Model
    ↓
 Parser
    ↓
Final Output
```

The output of one Runnable becomes the input of the next Runnable.

---

# 6. How does `|` work?

When we write:

```python
chain = prompt | model | parser
```

LangChain creates a **Runnable sequence**.

Conceptually:

```text
prompt
  │
  │ output
  ▼
model
  │
  │ output
  ▼
parser
```

So:

```python
chain.invoke(input)
```

roughly means:

```python
output1 = prompt.invoke(input)

output2 = model.invoke(output1)

output3 = parser.invoke(output2)

return output3
```

This is a simplified conceptual representation, but it captures the core idea.

---

# 7. How are Runnables implemented?

A component becomes a Runnable by implementing the Runnable abstraction/contract.

Conceptually:

```python
from langchain_core.runnables import Runnable


class MyRunnable(Runnable):

    def invoke(self, input, config=None):
        return input.upper()
```

Now:

```python
runnable = MyRunnable()

result = runnable.invoke("hello")

print(result)
```

Output:

```text
HELLO
```

The important thing is that `MyRunnable` follows the expected Runnable interface.

# 9. Important Runnable Methods

Runnable provides standardized execution patterns.

Common ones include:

```python
.invoke()
```

Run one input.

```python
.batch()
```

Run multiple inputs.

```python
.stream()
```

Stream output as it becomes available.

There are also asynchronous versions:

```python
.ainvoke()
.abatch()
.astream()
```

So Runnable is not only about `invoke()`.

It gives components a **standardized execution model**.

---

# 10. Why is this useful?

Without a common abstraction, we could end up with:

```text
Prompt → format()
Model → predict()
Parser → parse()
Retriever → retrieve()
```

Different components would have different execution interfaces.

With Runnable, LangChain can provide a more consistent model:

```text
Prompt      → invoke()
Model       → invoke()
Parser      → invoke()
Retriever   → invoke()
```

This makes components easier to:

* compose
* reuse
* replace
* execute
* stream
* batch
* run asynchronously
